```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- DATA (0 = not rated) ----------
movies = ["Avengers", "Avatar", "Titanic", "Inception", "Frozen"]
A = np.array([
    [5, 4, 0, 5, 0],
    [4, 5, 0, 4, 0],
    [0, 0, 5, 0, 4],
    [0, 0, 4, 5, 5],
    [5, 4, 0, 0, 0],
], dtype=float)

# ---------- CHOOSE USER AND RANK HERE ----------
user_number = 1      # 1 to 5
k = 2                # 1, 2 or 3
u = user_number - 1

# ---------- SVD:  A = U Σ Vt ----------
U, S, Vt = np.linalg.svd(A, full_matrices=False)
print("Singular values:", np.round(S, 2))

# ---------- LOW-RANK APPROXIMATION (first k singular values) ----------
Ak = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
print("\nLow-rank approximation A_k (rank", k, "):")
print(np.round(Ak, 2))

# ---------- PREDICTION for unrated movies ----------
unrated = [j for j in range(len(movies)) if A[u, j] == 0]
names = [movies[j] for j in unrated]
scores = [Ak[u, j] for j in unrated]

print("\nUser", user_number, "- Predicted SVD Scores:")
for n, s in zip(names, scores):
    print(" ", n, ":", round(s, 2))

best = names[int(np.argmax(scores))]
print("\nRecommended Movie:", best)
print("Reason: highest predicted SVD score among the movies this user has not rated.")

# ---------- HEATMAPS (side by side) ----------
def draw(ax, M, title, fmt):
    im = ax.imshow(M, cmap="viridis", vmin=0, vmax=5)
    ax.set_xticks(range(5)); ax.set_xticklabels(movies, rotation=45)
    ax.set_yticks(range(5)); ax.set_yticklabels([f"User {i+1}" for i in range(5)])
    ax.set_title(title)
    for i in range(5):
        for j in range(5):
            ax.text(j, i, fmt.format(M[i, j]), ha="center", va="center",
                    color="white" if M[i, j] < 3 else "black", fontsize=9)
    return im

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
im = draw(axes[0], A, "Original User-Movie Ratings (0 = not rated)", "{:.0f}")
draw(axes[1], Ak, f"SVD Low-Rank Approximation (rank {k})", "{:.2f}")
fig.colorbar(im, ax=axes, label="Rating / score")
plt.savefig("heatmaps.png", dpi=150, bbox_inches="tight")
plt.show()

# ---------- BAR CHART of predicted scores ----------
plt.figure(figsize=(6, 4))
bars = plt.bar(names, scores, color="tab:blue")
plt.bar_label(bars, labels=[f"{s:.2f}" for s in scores])
plt.title(f"User {user_number}: Predicted SVD Score (unrated movies)")
plt.ylabel("Predicted SVD Score")
plt.savefig("prediction_chart.png", dpi=150, bbox_inches="tight")
plt.show()
```
